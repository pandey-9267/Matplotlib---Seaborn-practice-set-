```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# this file is exist in the seaborn own dataset
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

**HISTPLOT / DISPLOT**

```python
# by default bin value is 20 
plt.subplot(1,2,1)
sns.histplot(df['total_bill'], bins = 30, kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'], kde = True)
```

**JOINT PLOT**

```python
# kind is by deafult scatter
sns.jointplot(data = df, x = 'total_bill', y = 'tip', kind = 'hex')
```

```python
sns.jointplot(data = df, x = 'total_bill', y = 'tip', kind = 'reg')
```

**PAIR PLOT**

```python
sns.pairplot(df, hue = 'sex')
```

```python
sns.pairplot(df, hue = 'size', palette = 'rainbow')
```

**RUGPLOT**

```python
sns.rugplot(df)
```
