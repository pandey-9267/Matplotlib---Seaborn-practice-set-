```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**MATPLOTLIB BASICS**

```python
x = np.linspace(0,5,11)
```

```python
x
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

**Subplots in Matplotlib**

```python
plt.subplot(2,2,1)
plt.plot(x**4,x**3)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

**OBJECT ORIENTED WAY**

```python
fig = plt.figure()
plot1 = fig.add_axes([0.1,0.1,1,1])
plot1.plot(x,y)
plot2 = fig.add_axes([0.3,0.6,0.3,0.3])
plot2.plot(x,y)
plot3 = fig.add_axes([0.7,0.2,0.3,0.3])
plot3.plot(x,y)
```

**Types of Plots**

```python
plt.scatter(x,y)
```

```python
marks = [45, 50, 55, 60, 60, 65, 70, 75, 80, 85, 90]
plt.hist(marks)
```

```python
plt.boxplot(x)
```

**Saving Plots**

```python
plot = plt.figure()
ax = plot.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
